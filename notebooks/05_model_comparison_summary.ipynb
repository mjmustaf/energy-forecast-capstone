# Model Comparison Summary

summarization and comparison the performance of four different models trained to predict solar energy output.

## Performance Metrics
We compare the following models:
- Linear Regression
- Random Forest
- XGBoost
- K-Nearest Neighbors

```python
# Summary of RMSE and R² scores
import pandas as pd

results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest', 'XGBoost', 'K-Nearest Neighbors'],
    'RMSE': [39.41, 9.00, 9.92, 14.04],
    'R² Score': [-1.27, 0.88, 0.86, 0.72]
})

results.sort_values(by='R² Score', ascending=False, inplace=True)
results.reset_index(drop=True, inplace=True)
results
```

##  Observations
- Random Forest performed the best with an R² of 0.88.
- XGBoost was a close second.
- KNN performed ok, but not as well as the ensemble models.
- Linear Regression performed poorly, which can be due to non relationships.

```python
# Visualize comparison
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
sns.barplot(x='Model', y='R² Score', data=results)
plt.title('Model R² Score Comparison')
plt.ylabel('R² Score')
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```
